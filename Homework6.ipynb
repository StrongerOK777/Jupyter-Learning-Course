### 文件操作作业

#### 第1题：从键盘输入一些字符，逐行把它们写到txt文件，直到输入一个 @ 为止

要求：
- txt文件名称不限
- 每次输入的内容在txt内保存为一行
- 采用UTF-8编码格式

```python
file_name = "input_lines.txt"

with open(file_name, "w", encoding="utf-8") as f:
    while True:
        line = input("请输入一行内容（输入@结束）：")
        if line == "@":
            break
        f.write(line + "\n")

print(f"已保存到文件：{file_name}")
```

#### 第2题：生成学生成绩并进行统计分析
请编写一个 Python 程序，模拟 100 个学生的考试成绩，并将其写入一个 CSV 文件

要求：

- 共生成 100 名学生，姓名格式为 学生1 到 学生100；

- 每门课程的成绩使用正态分布生成，满足以下条件：

    - 语文：均值 75，标准差 10

    - 数学：均值 78，标准差 12

    - 英语：均值 72，标准差 8

- 所有成绩范围限制在 [0, 100]（即如果生成超出范围，需截断处理）


- 计算每个学生的总分；按照总分从高到低排序；

- 将排序后的结果写入一个名为 student_scores.csv 的文件，格式如下：

|姓名|语文|数学|英语|总分|
|-|-|-|-|-|
|学生35|88|93|85|266|


- 输出以下信息（打印到终端）：

    - 每门课程（语文、数学、英语）的最高分、最低分、平均分

    - 所有学生总分的最高分、最低分、平均分


```python
import random
import csv

random.seed(42) 

def clipped_normal(mean, std, n):
    scores = []
    for _ in range(n):
        s = round(random.gauss(mean, std))
        s = max(0, min(100, s))
        scores.append(s)
    return scores

n_students = 100
students = [f"学生{i}" for i in range(1, n_students + 1)]
chinese_scores = clipped_normal(75, 10, n_students)
math_scores = clipped_normal(78, 12, n_students)
english_scores = clipped_normal(72, 8, n_students)

records = []
for i in range(n_students):
    total = chinese_scores[i] + math_scores[i] + english_scores[i]
    records.append({
        "姓名": students[i],
        "语文": chinese_scores[i],
        "数学": math_scores[i],
        "英语": english_scores[i],
        "总分": total
    })

# 按总分从高到低排序
records.sort(key=lambda x: x["总分"], reverse=True)

# 写入CSV文件
csv_file = "student_scores.csv"
with open(csv_file, "w", newline="", encoding="utf-8-sig") as f:
    writer = csv.DictWriter(f, fieldnames=["姓名", "语文", "数学", "英语", "总分"])
    writer.writeheader()
    writer.writerows(records)

def stats(nums):
    return max(nums), min(nums), sum(nums) / len(nums)

c_max, c_min, c_avg = stats(chinese_scores)
m_max, m_min, m_avg = stats(math_scores)
e_max, e_min, e_avg = stats(english_scores)
t_max, t_min, t_avg = stats([r["总分"] for r in records])

print(f"语文 -> 最高分: {c_max}, 最低分: {c_min}, 平均分: {c_avg:.2f}")
print(f"数学 -> 最高分: {m_max}, 最低分: {m_min}, 平均分: {m_avg:.2f}")
print(f"英语 -> 最高分: {e_max}, 最低分: {e_min}, 平均分: {e_avg:.2f}")
print(f"总分 -> 最高分: {t_max}, 最低分: {t_min}, 平均分: {t_avg:.2f}")
print(f"\n排序后的成绩已保存到：{csv_file}")
```

#### 第3题：采用 faker 模块随机产生20个虚拟人物的资料，然后将姓名、性别、年龄、邮箱、手机存到一个 excel 表里面

excel表的格式为：

|姓名  |  性别 |  年龄 | 邮箱 | 手机|
|-|-|-|-|-|
|张三|...|...|...|...|


```python
from faker import Faker
import pandas as pd

# 使用中文地区，便于生成中文姓名
fake = Faker("zh_CN")

records = []
for _ in range(20):
    gender = fake.random_element(elements=("男", "女"))
    records.append({
        "姓名": fake.name_male() if gender == "男" else fake.name_female(),
        "性别": gender,
        "年龄": fake.random_int(min=18, max=60),
        "邮箱": fake.email(),
        "手机": fake.phone_number(),
    })

# 生成表格并导出到 Excel
df_people = pd.DataFrame(records, columns=["姓名", "性别", "年龄", "邮箱", "手机"])
excel_file = "faker_people.xlsx"
df_people.to_excel(excel_file, index=False)

print(f"已生成 Excel 文件：{excel_file}")
print(df_people.head())
```
