### 第1题

- 某小店，售卖咖啡一杯30元；一次性购买5杯打9折，一次性购买8杯打8折，一次性购买12杯以上打7折。
- 根据单价price与杯数ncup，计算总共价钱

```python
price = 30
ncup = int(input("输入杯数"))
total = 0
```

```python
# ========answer=======
if ncup < 5:
    total = price * ncup
elif ncup < 8:
    total = price * ncup * 0.9
elif ncup < 13:
    total = price * ncup * 0.8
else:
    total = price * ncup * 0.7

print("总价是", total)
```

### 第2题

- 连续输入若干个非0整数，当输入整数为0时，停止输入
- 计算输入整数的平均值、最大值、最小值、中位数

```python
# ========answer=======
ls = []
a, cnt, tot, mx, mn, md = 0,0,0,0,0,0
while True:
    a=int(input("输入一个整数，输入0结束"))
    if a == 0:
        break
    cnt += 1
    ls.append(a)
    tot += a
    mx = mx if mx > a else a
    mn = mn if mn < a else a

ls.sort()
if cnt % 2 ==0:
    md = (ls[cnt//2-1] + ls[cnt//2]) / 2
else:
    md = ls[cnt//2]

print("平均数是", tot/cnt)
print("最大数是", mx)
print("最小数是", mn)
print("中位数是", md)
```

### 第3题 字符统计

- 有一串字符，请统计出字母大写/字母小写/数字/其他符号各多少
- 输出结果

```python
string = "I am a boy. I am 22 years old. How old are you?"
# ========answer=======
l = len(string)
C,c,n,o=0,0,0,0
for i in range(l):
    if string[i] >='A' and string[i] <='Z':
        C += 1
    elif string[i] >='a' and string[i] <='z': 
        c += 1
    elif string[i] >='0' and string[i] <='9':
        n += 1
    else:
        o +=1

print("大写字母有", C)
print("小写字母有", c) 
print("数字有", n)
print("其他字符有", o)

```

### 第4题 计算下面表达式结果

$S=1+\frac{1}{2}+\frac{1}{4}+\frac{1}{7}+\frac{1}{11}+\frac{1}{16}+\frac{1}{22}+\frac{1}{29}+...$

- 要求当某一项小$10^{-6}$时，停止求和

```python
S = 0
```

```python
# ========answer=======
current, step = 1, 1
while 1000000/current >= 1:
    S += 1000000/current
    current += step
    step += 1

S /= 1000000
print(f"总和{S}\n一直乘到{step-1}项\n")

```
