#### 第1题：生成斐波那契数列
给定整数n，返回一个包含前n个斐波那契数的列表。

例如，n=5，输出应为[0, 1, 1, 2, 3]。

```python
fib = [0,1,1]
n=int(input("输入n的值"))

for i in range(3,n):
    fib.append(fib[i-1]+fib[i-2])

print(fib[:n])
```

#### 第2题：先随机产生20个大于1且小于10000的正整数，再将其中的质数从列表中去除

```python
import random
random_list = [random.randint(1,10000) for i in range(20)]

def prime(x):
    for i in range(2, int(x**0.5)+1):
        if x % i == 0:
            return False
    return True

for num in random_list:
    if prime(num) == False:
        random_list.remove(num);

print(random_list);
        
```

#### 第3题：密码检查，输入一个字符串，检查是否符合规定：
 - 包含数字
 - 包含大写字母
 - 包含小写字母
 - 大于7位
 - 包含特殊字符（~ @ # % ! & $）
 
以上必须至少包含3条，才是一个符合要求的密码

```python
pwd = input("请输入密码：")

# 五条规则：数字、大写、小写、长度>7、特殊字符
rules = 0

if any(ch.isdigit() for ch in pwd):
    rules += 1
if any(ch.isupper() for ch in pwd):
    rules += 1
if any(ch.islower() for ch in pwd):
    rules += 1
if len(pwd) > 7:
    rules += 1
if any(ch in "~@#%!&$" for ch in pwd):
    rules += 1

if rules >= 3:
    print(f"密码合格（满足 {rules}/5 条规则）")
else:
    print(f"密码不合格（仅满足 {rules}/5 条规则）")
```

#### 第4题：随机生成20个1-10范围内的整数，构建一个字典，keys是整数，values是该整数出现的次数

```python
import random

nums = [random.randint(1, 10) for i in range(20)]
count_dict = {}

for x in nums:
    count_dict[x] = count_dict.get(x, 0) + 1

print("随机列表:", nums)
print("计数字典:", count_dict)
```
