#### 第1题：用 random 模块随机生成一个三角形，用 math 模块判断其是否为锐角、直角、钝角三角形
**要求：**
- 随机生成三角形的三个边长（1-20内随机整数），并输出三个边长的值
- 首先判断三个边长能否组成三角形，如果不能，输出“无法组成三角形”；如果可以，输出“可以组成三角形”
- 再判断该三角形是否为锐角、直角、钝角三角形，并输出“锐角三角形”或“直角三角形”或“钝角三角形”

```python
from random import *

a = randint(1,20)
b = randint(1,20)
c = randint(1,20)

if b>c:
    b,c=c,b
if a>b:
    a,b=b,a
if b>c:
    b,c=c,b

print(f"a,b,c={a},{b},{c}\n")

if a+b<=c:
    print("无法组成三角形\n")
else:
    print("可以组成三角形\n")
    if a*a+b*b>c*c:
        print("锐角三角形")
    elif a*a+b*b==c*c:
        print("直角三角形")
    else:
        print("钝角三角形")
```

#### 第2题：zhdate 是一个与农历日期有关的第三方模块，请使用该模块及 calendar 模块生成一个假期安排表：
 - 输入年份
 - 计算出当年元旦、春节、清明节、劳动节、端午节、中秋节、国庆节
 - 各个节日的日期和假期长度按如下简化条件确定：
     - 元旦：1月1日，1天
     - 春节：除夕 - 大年初六，7天
     - 清明：4月5日，1天
     - 劳动节：5月1日，1天
     - 端午节：农历五月初五，1天
     - 中秋节：农历八月十五，1天
     - 国庆节：10月1日到3日，3天
 - 按照如下格式输出：
 
 节日      YYYY.MM.DD ~ YYYY.MM.DD      星期X ~ 星期X

```python
from zhdate import ZhDate
from datetime import date, timedelta
import calendar

year = int(input("请输入年份："))

weekdays = ["星期一", "星期二", "星期三", "星期四", "星期五", "星期六", "星期日"]

def weekday_cn(d):
    return weekdays[calendar.weekday(d.year, d.month, d.day)]

def fmt_date(d):
    return d.strftime("%Y.%m.%d")

# 固定公历节日
new_year_start = date(year, 1, 1)
new_year_end = new_year_start

qingming_start = date(year, 4, 5)
qingming_end = qingming_start

labor_start = date(year, 5, 1)
labor_end = labor_start

national_start = date(year, 10, 1)
national_end = date(year, 10, 3)

# 农历节日
spring_festival_day1 = ZhDate(year, 1, 1).to_datetime().date()
spring_start = spring_festival_day1 - timedelta(days=1)   # 除夕
spring_end = spring_festival_day1 + timedelta(days=5)     # 大年初六

duanwu_start = ZhDate(year, 5, 5).to_datetime().date()
duanwu_end = duanwu_start

mid_autumn_start = ZhDate(year, 8, 15).to_datetime().date()
mid_autumn_end = mid_autumn_start

holidays = [
    ("元旦", new_year_start, new_year_end),
    ("春节", spring_start, spring_end),
    ("清明节", qingming_start, qingming_end),
    ("劳动节", labor_start, labor_end),
    ("端午节", duanwu_start, duanwu_end),
    ("中秋节", mid_autumn_start, mid_autumn_end),
    ("国庆节", national_start, national_end),
]

print("节日     YYYY.MM.DD ~ YYYY.MM.DD      星期几 ~ 星期几")
for name, start, end in holidays:
    if name in ["元旦","春节"]:
        print(
        f"{name:<4}   {fmt_date(start)} ~ {fmt_date(end)}      "
        f"{weekday_cn(start)} ~ {weekday_cn(end)}"
    )
    else:
        print(
            f"{name:<4}  {fmt_date(start)} ~ {fmt_date(end)}      "
            f"{weekday_cn(start)} ~ {weekday_cn(end)}"
        )
```
